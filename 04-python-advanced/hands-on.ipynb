## CLASS BASICS & ATTRIBUTES

```python
# TODO 1: Buat class Mobil dengan:
# - Class attribute: jumlah_roda = 4
# - Instance attributes: merk, warna, tahun (via __init__)
# - Method info() yang return informasi mobil

class Mobil:
    # Class attribute
    jumlah_roda = 4

    def __init__(self, merk, warna, tahun):
        # Instance attributes
        self.merk = merk
        self.warna = warna
        self.tahun = tahun

    def info(self): #method pertama
        return f"Mobil {self.merk} warna {self.warna} tahun {self.tahun}"

# Testing
mobil_ayah = Mobil("Toyota", "Hitam", 2021)
mobil_adik = Mobil("Honda", "Merah", 2022)

print("\n--- Testing Mobil Class ---")
print(f"Class attribute (jumlah_roda): {Mobil.jumlah_roda}")
print(mobil_ayah.info())
print(mobil_adik.info())

# TODO 2: Tunjukkan bahwa class attribute shared
print("\n--- Class Attribute adalah Shared ---")
print(f"mobil_ayah.jumlah_roda: {mobil_ayah.jumlah_roda}")
print(f"mobil_adik.jumlah_roda: {mobil_adik.jumlah_roda}")

# Ubah via class
Mobil.jumlah_roda = 6
print(f"Setelah Mobil.jumlah_roda = 6:")
print(f"mobil_ayah.jumlah_roda: {mobil_ayah.jumlah_roda}")
print(f"mobil_adik.jumlah_roda: {mobil_adik.jumlah_roda}")
```

```python

print("\n" + "="*70)
print("EXERCISE 2: METHODS & SELF")
print("="*70)

# TODO 3: Buat class Lingkaran dengan method keliling dan luas

class Lingkaran:
    #inisiasi jari jari
    def __init__(self, jari_jari):
        self.jari_jari = jari_jari

    #method
    def keliling(self):
        return 2 * 3.14159 * self.jari_jari

    def luas(self):
        return 3.14159 * self.jari_jari ** 2

    def info(self):
        return f"Lingkaran dengan jari-jari {self.jari_jari} cm" # f untuk manggil var jari2

# Testing
lingkaran1 = Lingkaran(7)
lingkaran2 = Lingkaran(10)

print("\n--- Testing Lingkaran Class ---")
print(lingkaran1.info())
print(f"Keliling: {lingkaran1.keliling():.2f} cm")
print(f"Luas: {lingkaran1.luas():.2f} cm²")

print(f"\n{lingkaran2.info()}")
print(f"Keliling: {lingkaran2.keliling():.2f} cm")
print(f"Luas: {lingkaran2.luas():.2f} cm²")
```

## INHERITANCE

```python
# TODO 4: Implementasi inheritance Warga -> Satpam -> TNI
# Sesuai dengan contoh di slide

class Warga:
    # inisiasi parent
    def __init__(self, nama, nik):
        self.nama = nama
        self.nik = nik

    # method parent
    def perkenalan(self):
        return f"Nama: {self.nama}, NIK: {self.nik}"

    def nyoblos(self):
        return f"{self.nama} nyoblos di TPS"

class Satpam(Warga): #mewarisi dr warga

    #inisiasi
    def __init__(self, nama, nik, shift):
        super().__init__(nama, nik) # super: tolong kamu mewarisi dr parent panggil nama sm nik
        self.shift = shift
    #method
    def bertugas(self):
        return f"{self.nama} bertugas shift {self.shift}"

class TNI(Warga):
    def __init__(self, nama, nik, pangkat):
        super().__init__(nama, nik) # super: agar nama sm nik tdk perlu di panggil lagi
        self.pangkat = pangkat

    def bertugas(self):
        return f"{self.pangkat} {self.nama} siap bertugas"

# Testing
print("\n--- Testing Inheritance ---")
warga = Warga("Budi", "3201001")
satpam = Satpam("Mukhlis", "3201002", "Malam")
tni = TNI("Ahmad", "3201003", "Sersan")

print("\nWarga biasa:")
print(warga.perkenalan())
print(warga.nyoblos())

print("\nSatpam (inherit dari Warga):")
print(satpam.perkenalan())  # Method dari parent
print(satpam.nyoblos())     # Method dari parent
print(satpam.bertugas())    # Method sendiri

print("\nTNI (inherit dari Warga):")
print(tni.perkenalan())
print(tni.nyoblos())
print(tni.bertugas())
```

## POLYMORPHISM (Method Overriding)

```python
# TODO 5: Override method nyoblos di class TNI
# TNI tidak boleh nyoblos

class WargaV2:
    def __init__(self, nama, nik):
        self.nama = nama
        self.nik = nik

    def nyoblos(self):
        return f"{self.nama} nyoblos dengan resmi"

class TNIV2(WargaV2):
    def __init__(self, nama, nik, pangkat):
        super().__init__(nama, nik)
        self.pangkat = pangkat

    # OVERRIDE method nyoblos
    def nyoblos(self):
        return f"{self.nama} (TNI) tidak boleh nyoblos"

class SatpamV2(WargaV2):
    # Tidak override, pakai method parent
    pass

# Testing
print("\n--- Testing Polymorphism ---")
warga_v2 = WargaV2("Budi", "123")
tni_v2 = TNIV2("Ahmad", "456", "Sersan")
satpam_v2 = SatpamV2("Mukhlis", "789")

print(warga_v2.nyoblos())   # Method asli
print(tni_v2.nyoblos())     # Method yang di-override
print(satpam_v2.nyoblos())  # Method dari parent
```

## POLYMORPHISM (Different Classes, Same Method)

```python
# TODO 6: Buat Email, SMS, WhatsApp dengan method send()

class Email:
    def send(self, message):
        return f"📧 Sending via Email: {message}"

class SMS:
    def send(self, message):
        return f"📱 Sending via SMS: {message}"

class WhatsApp:
    def send(self, message):
        return f"💬 Sending via WhatsApp: {message}"

# Polymorphic function
def kirim_notifikasi(channel, message):
    print(channel.send(message))

# Testing
print("\n--- Testing Polymorphism (Duck Typing) ---")
email = Email()
sms = SMS()
whatsapp = WhatsApp()

kirim_notifikasi(email, "Meeting at 10 AM")
kirim_notifikasi(sms, "Meeting at 10 AM")
kirim_notifikasi(whatsapp, "Meeting at 10 AM")
```

## ABSTRACTION

```python
# TODO 7: Buat abstract class dengan abstract methods
# Sesuai contoh di slide

from abc import ABC, abstractmethod

#parent
class Kendaraan(ABC):
    def __init__(self, merk, tahun):
        self.merk = merk
        self.tahun = tahun

    #wajib mewarisi nyalakan mesin dan matikan mesin, isi bebas
    @abstractmethod
    def nyalakan_mesin(self):
        pass

    @abstractmethod
    def matikan_mesin(self):
        pass

    # Concrete method
    def info(self):
        return f"Kendaraan {self.merk} tahun {self.tahun}"

#child
class MobilKendaraan(Kendaraan):
    def nyalakan_mesin(self):
        return f"🚗 {self.merk}: Mesin mobil menyala dengan kunci"

    def matikan_mesin(self):
        return f"🚗 {self.merk}: Mesin mobil mati"

class Motor(Kendaraan):
    def nyalakan_mesin(self):
        return f"🏍️ {self.merk}: Mesin motor menyala dengan starter"

    def matikan_mesin(self):
        return f"🏍️ {self.merk}: Mesin motor mati"

# Testing
print("\n--- Testing Abstraction ---")
# kendaraan = Kendaraan("Generic", 2020)  # ❌ ERROR! Can't instantiate

mobil_kendaraan = MobilKendaraan("Toyota", 2020)
motor = Motor("Honda", 2021)

print(mobil_kendaraan.info())
print(mobil_kendaraan.nyalakan_mesin())
print(mobil_kendaraan.matikan_mesin())

print(f"\n{motor.info()}")
print(motor.nyalakan_mesin())
print(motor.matikan_mesin())

```

## ENCAPSULATION (Private Attributes)

```python
# TODO 8: Buat class dengan private attributes dan getter/setter

class BankAccount:
    def __init__(self, nama, saldo_awal):
        self.nama = nama              # Public
        self.__saldo = saldo_awal     # rahasia, private, __
        self.__pin = "1234"           # rahasia

    # Getter untuk private attribute
    # method untuk dapet akses saldo
    def get_saldo(self):
        return self.__saldo

    # Setter untuk private attribute (dengan validasi)
    def set_pin(self, pin_lama, pin_baru):
        if pin_lama == self.__pin:
            self.__pin = pin_baru
            return "PIN berhasil diubah"
        return "PIN lama salah!"

    def setor(self, jumlah):
        if jumlah > 0:
            self.__saldo += jumlah
            return f"✅ Setor Rp {jumlah:,} berhasil"
        return "❌ Jumlah tidak valid"

    def tarik(self, jumlah, pin):
        if pin != self.__pin:
            return "❌ PIN salah!"
        if jumlah > self.__saldo:
            return "❌ Saldo tidak cukup"
        if jumlah > 0:
            self.__saldo -= jumlah
            return f"✅ Tarik Rp {jumlah:,} berhasil"
        return "❌ Jumlah tidak valid"

# Testing
print("\n--- Testing Encapsulation ---")
akun = BankAccount("Budi", 1000000)

# Public attribute bisa diakses
print(f"Pemilik: {akun.nama}")

# Private attribute TIDAK bisa diakses langsung
# print(akun.__saldo)  # ❌ ERROR!

# Akses via getter
print(f"Saldo: Rp {akun.get_saldo():,}")

# Modifikasi via method
print(akun.setor(500000))
print(f"Saldo: Rp {akun.get_saldo():,}")

print(akun.tarik(300000, "1234"))
print(f"Saldo: Rp {akun.get_saldo():,}")

print(akun.tarik(2000000, "1234"))  # Saldo tidak cukup
print(akun.tarik(100000, "9999"))   # PIN salah

```

## PROPERTY DECORATOR (Pythonic Encapsulation)

```python
# TODO 9: Gunakan @property decorator

class Person:
    def __init__(self, nama, umur):
        self.__nama = nama
        self.__umur = umur

    # Property sebagai getter
    @property
    def nama(self):
        return self.__nama

    @nama.setter
    def nama(self, value):
        if len(value) > 0:
            self.__nama = value
        else:
            raise ValueError("Nama tidak boleh kosong")

    @property
    def umur(self):
        return self.__umur

    @umur.setter
    def umur(self, value):
        if 0 <= value <= 150:
            self.__umur = value
        else:
            raise ValueError("Umur tidak valid (0-150)")

    @property
    def info(self):
        return f"{self.__nama} berumur {self.__umur} tahun"

# Testing
print("\n--- Testing Property Decorator ---")
person = Person("Budi", 25)

# Akses seperti attribute (tapi lewat getter)
print(f"Nama: {person.nama}")
print(f"Umur: {person.umur}")

# Set seperti attribute (tapi lewat setter dengan validasi)
person.nama = "Ahmad"
person.umur = 30
print(person.info)

# Validasi otomatis
try:
    person.umur = -5  # Error!
except ValueError as e:
    print(f"Error: {e}")

try:
    person.umur = 200  # Error!
except ValueError as e:
    print(f"Error: {e}")
```

## COMPREHENSIVE EXERCISE: SISTEM RENTAL MOBIL

```python
# TODO 10: Gabungkan semua konsep OOP

from abc import ABC, abstractmethod
from datetime import datetime, timedelta

class Vehicle(ABC):
    """Abstract base class untuk semua kendaraan"""
    vehicle_count = 0  # Class attribute

    def __init__(self, vehicle_id, merk, tahun):
        self.vehicle_id = vehicle_id
        self.merk = merk
        self.tahun = tahun
        self._is_rented = False  # Protected
        Vehicle.vehicle_count += 1

    @abstractmethod
    def harga_sewa_per_hari(self):
        """Setiap kendaraan punya harga berbeda"""
        pass

    @property
    def is_rented(self):
        return self._is_rented

    @is_rented.setter
    def is_rented(self, status):
        self._is_rented = status

    def info(self):
        status = "Tersewa" if self._is_rented else "Tersedia"
        return f"[{self.vehicle_id}] {self.merk} ({self.tahun}) - {status}"

class Car(Vehicle):
    def __init__(self, vehicle_id, merk, tahun, jenis):
        super().__init__(vehicle_id, merk, tahun)
        self.jenis = jenis  # SUV, Sedan, MPV

    def harga_sewa_per_hari(self):
        harga_base = {"SUV": 500000, "Sedan": 350000, "MPV": 400000}
        return harga_base.get(self.jenis, 300000)

class Motorcycle(Vehicle):
    def __init__(self, vehicle_id, merk, tahun, cc):
        super().__init__(vehicle_id, merk, tahun)
        self.cc = cc

    def harga_sewa_per_hari(self):
        if self.cc <= 150:
            return 100000
        elif self.cc <= 250:
            return 150000
        else:
            return 200000

class RentalSystem:
    def __init__(self):
        self.__vehicles = []  # Private
        self.__rentals = []   # Private

    def add_vehicle(self, vehicle):
        self.__vehicles.append(vehicle)
        print(f"✅ Kendaraan ditambahkan: {vehicle.info()}")

    def list_vehicles(self):
        print("\n--- DAFTAR KENDARAAN ---")
        for v in self.__vehicles:
            harga = v.harga_sewa_per_hari()
            print(f"{v.info()} - Rp {harga:,}/hari")

    def rent_vehicle(self, vehicle_id, nama_penyewa, hari):
        for v in self.__vehicles:
            if v.vehicle_id == vehicle_id:
                if v.is_rented:
                    return "❌ Kendaraan sedang disewa"

                v.is_rented = True
                total = v.harga_sewa_per_hari() * hari
                rental = {
                    'vehicle': v,
                    'penyewa': nama_penyewa,
                    'hari': hari,
                    'total': total,
                    'tanggal': datetime.now()
                }
                self.__rentals.append(rental)
                return f"✅ Sewa berhasil! Total: Rp {total:,} untuk {hari} hari"

        return "❌ Kendaraan tidak ditemukan"

    def return_vehicle(self, vehicle_id):
        for rental in self.__rentals:
            if rental['vehicle'].vehicle_id == vehicle_id:
                rental['vehicle'].is_rented = False
                self.__rentals.remove(rental)
                return "✅ Kendaraan berhasil dikembalikan"
        return "❌ Kendaraan tidak sedang disewa"

# Testing Comprehensive
print("\n--- TESTING SISTEM RENTAL ---")
rental = RentalSystem()

# Add vehicles (Polymorphism - berbagai tipe kendaraan)
rental.add_vehicle(Car("C001", "Toyota Fortuner", 2022, "SUV"))
rental.add_vehicle(Car("C002", "Honda Civic", 2021, "Sedan"))
rental.add_vehicle(Motorcycle("M001", "Honda PCX", 2023, 150))
rental.add_vehicle(Motorcycle("M002", "Yamaha R25", 2022, 250))

# List semua kendaraan
rental.list_vehicles()

# Sewa kendaraan
print("\n--- PROSES PENYEWAAN ---")
print(rental.rent_vehicle("C001", "Budi", 3))
print(rental.rent_vehicle("M001", "Ahmad", 5))

# List lagi (status berubah)
rental.list_vehicles()

# Return kendaraan
print("\n--- PENGEMBALIAN ---")
print(rental.return_vehicle("C001"))

# List lagi
rental.list_vehicles()

print("\n" + "="*70)
print(f"Total Vehicle Created: {Vehicle.vehicle_count}")
print("="*70)
```
