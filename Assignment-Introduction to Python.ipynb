### 2. Soal 2: Membuat input

Buatlah dua kode agar pengguna bisa melakukan input untuk data berikut:

Nama produk

Jumlah produk terjual

Harga produk


Goals: Membuat kode agar pengguna dapat memasukkan input-an ke dalam program.


```python
nama_product = input('Masukkan Nama Product : ')
jumlah_product_terjual = int(input('Masukkan Jumlah Product Terjual : '))
harga_product = int(input('Masukkan Harga Product : '))
```

### 3. Soal 3: Logika If/Else untuk Rekomendasi Produk

Lakukan perhitungan total revenue dari penjualan barang dengan rumus:

total revenue = jumlah_produk_terjual  x harga_produk.

Gunakan conditional statement untuk melihat hasil akhir penilaian:

*   Jika total revenue kurang dari atau sama dengan 2.000.000, maka kategori penjualan “Rendah”.
*   Jika total revenue lebih dari 2.000.000 dan total revenue kurang dari atau sama dengan  8.000.000, maka kategori penjualan “Sedang”.

*   Jika total revenue lebih dari 8.000.000, maka kategori penjualan “Tinggi”.


Goals: Memahami cara if/else bekerja.








```python
total_rev = jumlah_product_terjual * harga_product

if total_rev <= 2000000:
  kategori = 'Rendah'
elif total_rev > 2000000 and total_rev <= 8000000:
  kategori = 'Sedang'
else:
  kategori = 'Tinggi'

print('Kategori Penjualan',kategori)
```

#### 4. Soal 4: Output Akhir Program

Goals: Menampilkan rekomendasi produk.



```python
print('Nama Product :', nama_product)
print('Jumlah Terjual :', jumlah_product_terjual)
print('Harga Product :', harga_product)
print('Total Revenue :',total_rev)
print('Kategori Penjualan :',kategori)
```
